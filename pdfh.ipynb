## Install and Use Camelot for Table Extraction
Camelot is a Python library specialized for extracting tables from PDFs. It supports both lattice-based and stream-based table detection.

```python
import camelot  # import the Camelot table extraction library for PDFs

lattice_tables = camelot.read_pdf(
    r".\data\tsla-20260630-gen.pdf", 
    pages="all", 
    flavor="lattice", 
    suppress_stdout=False
)
stream_tables = camelot.read_pdf(
    r".\data\tsla-20260630-gen.pdf", 
    pages="all", 
    flavor="stream",  
    suppress_stdout=False
)

# Print DataFrame output for all tables detected by lattice mode
for table in lattice_tables:
    print(table.df)

# Print DataFrame output for all tables detected by stream mode
for table in stream_tables:
    print(table.df)
```

## Extract Tables with pdfplumber
`pdfplumber` can detect table structures directly from PDF page contents by analyzing line and text positions.

```python
tables = []  # list to store table extraction results from all pages

with pdfplumber.open(r".\data\tsla-20260630-gen.pdf") as pdf:
    for page in pdf.pages:
        tables_on_page = page.extract_tables({})  # default extraction settings for tables

        if tables_on_page:
            for table in tables_on_page:
                tables.append({
                    "page": pdf.pages.index(page) + 1,  # record page number starting at 1
                    "data": table
                })

# Convert each extracted table to a pandas DataFrame and print it with its page number
for table in tables:
    print(f"Page {table['page']}: ")
    print(pd.DataFrame(table["data"]))
```

## Refine pdfplumber Table Extraction with Custom Strategies
For complex page layouts, specify custom extraction strategies and tolerance settings to better detect table boundaries.

```python
tables = []  # list to store refined extraction results

with pdfplumber.open(r".\data\tsla-20260630-gen.pdf") as pdf:
    for page in pdf.pages:
        tables_on_page = page.extract_tables({
            "vertical_strategy": "text",
            "horizontal_strategy": "text",
            "intersection_x_tolerance": 10,
            "intersection_y_tolerance": 10
        })

        if tables_on_page:
            for table in tables_on_page:
                tables.append({
                    "page": pdf.pages.index(page) + 1,
                    "data": table
                })

# Print each table with its source page after refined extraction
for table in tables:
    print(f"Page {table['page']}: ")
    print(pd.DataFrame(table["data"]))
```

## Extract Text and Metadata with PyPDF2
`PyPDF2` is a lightweight PDF library used to read text and metadata from PDF documents.

```python
import PyPDF2  # PDF reading and metadata extraction library

text = ""  # accumulate extracted text from all pages

with open(r".\data\tsla-20260630-gen.pdf", "rb") as file:
    reader = PyPDF2.PdfReader(file)
    num_pages = len(reader.pages)

    for page_num in range(num_pages):
        page = reader.pages[page_num]
        page_text = page.extract_text()  # extract text from the current page
        text += page_text + '\n\n'  # add a blank line between pages

print(text)
```

## Discover Images inside the PDF
This example enumerates embedded images in the PDF pages and prints image metadata such as dimensions and format.

```python
from pathlib import Path

pdf_path = Path("./data/tsla-20260630-gen.pdf")

with pdf_path.open("rb") as file:
    reader = PyPDF2.PdfReader(file)
    print(f"Number of pages: {len(reader.pages)}")

    total_images = 0

    for page_number, page in enumerate(reader.pages, start=1):
        page_images = page.images  # list of image objects extracted from the page

        if not page_images:
            continue

        print(f"\n--- Page {page_number} ({len(page_images)} image(s)) ---")
        for img in page_images:
            total_images += 1
            print(
                f"Image: {img.name} | Name: {img.name} | "
                f"Dimensions: {img.image.width}x{img.image.height} | "
                f"Format: {img.image.format}"
            )

print(f"\nTotal images found across document: {total_images}")
```

## Inspect PDF with PyMuPDF (`fitz`)
`PyMuPDF` offers advanced access to PDF metadata, page content, and document structure.

```python
import fitz  # PyMuPDF

print(fitz.__doc__)  # print library documentation summary for the current fitz installation
```

```python
with fitz.open(r".\data\tsla-20260630-gen.pdf") as pdf:
    print(f"Number of pages: {pdf.page_count}")
    print(f"Metadata: {pdf.metadata}")
    print(f"Table of Contents: {pdf.get_toc()}")

    page = pdf.load_page(0)  # load the first page of the document
    text = page.get_text("text")  # extract plain text from the first page
    with open("page1_text.txt", "w", encoding="utf-8") as f:
        clean_text = text.replace("\t", " ")  # normalize tab characters to spaces
        f.write("\n--- Page 1 Text ---\n" + clean_text)
```
